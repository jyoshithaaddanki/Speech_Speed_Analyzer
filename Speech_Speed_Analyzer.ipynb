```python
# 🎤 SPEECH SPEED ANALYZER - NO FILE NEEDED

!pip install librosa matplotlib -q

import numpy as np
import librosa
import librosa.display
import matplotlib.pyplot as plt
from scipy.io.wavfile import write
from IPython.display import Audio

print("🎤 SPEECH SPEED ANALYZER")
print("========================")

# Create sample audio automatically
sample_rate = 22050
duration = 10

t = np.linspace(0, duration, sample_rate * duration)

# Create a sample audio signal
audio = 0.3 * np.sin(2 * np.pi * 440 * t)

# Save audio file
filename = "sample_speech.wav"
write(filename, sample_rate, audio.astype(np.float32))

print("\nSample audio created successfully!")
print("File:", filename)

# Play sample audio
display(Audio(filename))

# Analyze audio
audio, sr = librosa.load(filename, sr=None)

# Duration
total_duration = len(audio) / sr

# Volume analysis
rms = librosa.feature.rms(y=audio)[0]

average_volume = np.mean(rms)

# Estimate speed
if total_duration < 5:
    speed = "Fast"
elif total_duration < 12:
    speed = "Normal"
else:
    speed = "Slow"

print("\n===== SPEECH ANALYSIS =====")
print("File:", filename)
print("Sample Rate:", sr, "Hz")
print("Duration:", round(total_duration, 2), "seconds")
print("Average Volume:", round(average_volume, 4))
print("Speech Speed:", speed)

# Waveform
plt.figure(figsize=(10, 4))

librosa.display.waveshow(audio, sr=sr)

plt.title("Speech Waveform")
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.grid()

plt.show()
```
